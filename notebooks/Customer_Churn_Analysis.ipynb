```python
import pandas as pd

df = pd.read_csv("dataset/WA_Fn-UseC_-Telco-Customer-Churn.csv")

print(df.shape)
print(df.head())
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df = df.dropna(subset=["TotalCharges"])

df = df.drop("customerID", axis=1)

df["Churn"] = df["Churn"].map({"No": 0, "Yes": 1})

print(df.shape)
```

```python
X = df.drop("Churn", axis=1)
y = df["Churn"]

print("X:", X.shape)
print("y:", y.shape)
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)
X_encoded = X_encoded.astype(int)

print("Original shape:", X.shape)
print("Encoded shape:", X_encoded.shape)
```

```python
import numpy as np

np.random.seed(42)

indices = np.random.permutation(len(X_encoded))
split = int(len(indices) * 0.80)

train_indices = indices[:split]
test_indices = indices[split:]

X_train = X_encoded.iloc[train_indices]
X_test = X_encoded.iloc[test_indices]

y_train = y.iloc[train_indices]
y_test = y.iloc[test_indices]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from xgboost import XGBClassifier

model = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="logloss"
)

model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("First 20 predictions:")
print(y_pred[:20])
```

```python
accuracy = (y_pred == y_test).mean()

print(f"Accuracy: {accuracy:.2%}") 
```

```python
from collections import Counter

tp = ((y_test == 1) & (y_pred == 1)).sum()
tn = ((y_test == 0) & (y_pred == 0)).sum()
fp = ((y_test == 0) & (y_pred == 1)).sum()
fn = ((y_test == 1) & (y_pred == 0)).sum()

print("True Positives :", tp)
print("True Negatives :", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
```

```python
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

print(f"Precision: {precision:.2%}")
print(f"Recall:    {recall:.2%}")
print(f"F1 Score:  {f1:.2%}")
```

```python
negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

scale_pos_weight = negative / positive

print("Customers who stayed:", negative)
print("Customers who churned:", positive)
print("Scale position weight:", scale_pos_weight)
```

```python
from xgboost import XGBClassifier

improved_model = XGBClassifier(
    n_estimators=300,
    max_depth=3,
    learning_rate=0.03,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=2.8032,
    random_state=42,
    eval_metric="logloss"
)

improved_model.fit(X_train, y_train)

print("Improved model training completed!")
```

```python
improved_pred = improved_model.predict(X_test)

print("Predictions completed!")
```

```python
tp = ((y_test == 1) & (improved_pred == 1)).sum()
tn = ((y_test == 0) & (improved_pred == 0)).sum()
fp = ((y_test == 0) & (improved_pred == 1)).sum()
fn = ((y_test == 1) & (improved_pred == 0)).sum()

accuracy = (improved_pred == y_test).mean()
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

print(f"Accuracy:  {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall:    {recall:.2%}")
print(f"F1 Score:  {f1:.2%}")
print()
print("True Positives :", tp)
print("True Negatives :", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
```

```python
importance = improved_model.feature_importances_

feature_importance = pd.DataFrame({
    "Feature": X_encoded.columns,
    "Importance": importance
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance.head(10))
```
